```python
import numpy as np
import pandas as pd
arr = np.array([10,70,40,66,33])
print(arr)

sum = arr.sum()
avg = sum/5
print("Sum of marks: ", sum)
print("Average:",avg)

highest_marks = np.max(arr)
print("Highest marks: ", highest_marks)

lowest_marks = np.min(arr)
print("Lowest marks: ", lowest_marks)

names = np.array(["Jawad", "Araiz", "Abdullah", "Hanzla", "Mubashir"])

data = {
    "Names": names,
    "Marks": arr
}

df = pd.DataFrame(data)
print(df)

```

```python
import matplotlib.pyplot as plt
plt.bar(df["Names"], df["Marks"])

plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```
