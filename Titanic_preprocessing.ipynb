```python
import pandas as pd
df = pd.read_csv("/content/drive/MyDrive/Titanic-Dataset.csv")
```

```python
print("First 5 rows of the dataset: ")
print(df.head())
```

```python
print("Data Description: ")
print(df.info())
```

```python
print("Missing values: ")
print(df.isnull().sum)
```

```python
df = df.drop(columns=["PassengerId", "Name", "Ticket", "Cabin"])
```

```python
# 6. Fill missing Age values with the median
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
# 7. Fill missing Embarked values with the most common value
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# 8. Convert Sex into numerical values
df["Sex"] = df["Sex"].map({
    "male": 0,
    "female": 1
})

# 9. Convert Embarked into numerical values
df["Embarked"] = df["Embarked"].map({
    "S": 0,
    "C": 1,
    "Q": 2
})

# 10. Check missing values again
print("\nMissing Values After Preprocessing:")
print(df.isnull().sum())

# 11. Display the processed dataset
print("\nProcessed Dataset:")
print(df.head())

# 12. Display data types
print("\nData Types:")
print(df.dtypes)

# 13. Save the processed dataset
df.to_csv("titanic_processed.csv", index=False)

print("\nProcessed dataset saved as titanic_processed.csv")
```
